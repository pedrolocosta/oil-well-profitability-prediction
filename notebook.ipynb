## Encapsulating the Approach in a Function

```python
def train_and_evaluate(df, region, random_state=12345):

    '''Train a linear regression model and evaluate it on a validation set.'''

    # Separate the feature columns from the target column
    features = df.drop(['id','product_in_thousands'], axis=1)
    target = df['product_in_thousands']

    # Split the data into training and validation sets using a 75:25 ratio
    features_train, features_valid, target_train, target_valid = train_test_split(
        features, target, test_size=0.25, random_state=random_state)
    
    # Fit the scaler using only the training set
    scaler = StandardScaler().fit(features_train)

    # Initialize and train the model
    model = LinearRegression()
    model.fit(scaler.transform(features_train), target_train)

    # Predict validation reserves while preserving the original index to align predictions with actual values
    predictions_valid = pd.Series(
        model.predict(scaler.transform(features_valid)),
        index=target_valid.index,
        name='predictions_valid',
    )

    # Organize the validation data in a DataFrame
    df_lr_valid = features_valid.copy()
    df_lr_valid['target_valid'] = target_valid
    df_lr_valid['predictions_valid'] = predictions_valid
    
    # Evaluate model performance
    # Calculate the actual mean reserves for the region
    mean_region = df['product_in_thousands'].mean()
    # Calculate the mean predicted reserves
    predictions_valid_mean = predictions_valid.mean()
    # Calculate the model RMSE
    rmse = mean_squared_error(target_valid, predictions_valid) ** 0.5
    # Calculate the coefficient of determination
    r2 = r2_score(target_valid, predictions_valid)
    # Calculate the standard deviations
    target_valid_std = target_valid.std()
    predictions_valid_std = predictions_valid.std()
    
    # Display the validation results
    print(f'Model performance evaluation for Region {region}:\n')
    print(f'Actual mean reserves for region {region}: {mean_region:.2f} thousand barrels')
    print(f'Mean predicted reserves: {predictions_valid_mean:.2f} thousand barrels')
    print(f'Model RMSE: {rmse:.2f}')
    print(f'Model R²: {r2:.2f}')
    print(
        f'Standard deviation of actual values: {target_valid_std:.2f}', 
        f'| Standard deviation of predicted values {predictions_valid_std:.2f}'
    )
    print('\nValidation DataFrame:\n', df_lr_valid.head())

    return target_valid, predictions_valid
    
```

## Applying the Function to Each Region

### Region 00

```python
# Apply the training and evaluation function to region 00
target_valid_0, predictions_valid_0 = train_and_evaluate(df0, '00')
```

### Region 01

```python
# Apply the training and evaluation function to region 01
target_valid_1, predictions_valid_1 = train_and_evaluate(df1, '01')
```

### Region 02

```python
# Apply the training and evaluation function to region 02
target_valid_2, predictions_valid_2 = train_and_evaluate(df2, '02')
```

### Grouping the Results

```python
# Group the prediction results for each region
regions = {
    '00' : (target_valid_0, predictions_valid_0),
    '01' : (target_valid_1, predictions_valid_1),
    '02' : (target_valid_2, predictions_valid_2),
}
```

# Calculating Profit from the Best Oil Wells

## Creating a Function to Calculate Profit Across Regions

```python
def profit(target, predictions, count=total_well):

    '''Calculate profit for the `count` wells with the highest model predictions.'''

    # Get the indices of the wells with the highest predicted reserves
    best_wells = predictions.sort_values(ascending=False).head(count)

    # Get the actual reserves of these wells (thousand barrels)
    selected = target.loc[best_wells.index]

    # Profit = revenue from the selected wells - investment
    return selected.sum() * revenue_uni - investiment
```

## Applying the Function to Each Region

```python
profits = {}
top_predictions = {}
for area, (target_valid, predictions_valid) in regions.items():
    # Calculate profit for the 200 wells with the highest predictions in each region
    profits[area] = profit(target_valid, predictions_valid)
    
    # Select the 200 highest well predictions for each region
    top_predictions[area] = predictions_valid.sort_values(ascending=False).head(total_well)
    
    # Get the actual reserves of the 200 selected wells in each region
    real_volume = target_valid.loc[top_predictions[area].index].sum()
    
    # Calculate the mean reserves per selected well in each region
    mean_well = target_valid.loc[top_predictions[area].index].mean()

    # Display a summary of the results
    print(f'Region {area}:', '\n',
          f'Mean reserves per well (thousand barrels) {mean_well:.2f}', '\n',
          f'Total production volume = {real_volume:,.0f} thousand barrels', '\n',
          f'Potential profit = US$ {profits[area]:,.0f}', '\n')
```

## Evaluating the Profit Estimates

**Potential profit assessment: the 200 wells with the highest predicted reserves**

Region 00 has the highest potential profit (USD 33.2 million), followed by Region 02 (USD 27.1 million) and Region 01 (USD 24.2 million). In every region, the mean actual reserves of the 200 selected wells are well above the break-even threshold of 111.1 thousand barrels per well: approximately 148 thousand barrels in Region 00, 141 thousand in Region 02, and 138 thousand in Region 01. This shows that selecting the best wells using the model can make the project profitable even when regional means are below the break-even threshold.

Region 01 has the lowest profit but a distinctive characteristic: its 200 selected wells all have the region's maximum reserve volume (137.9 thousand barrels). This is therefore close to the highest possible result for the region, achieved because the model makes almost no errors. In Regions 00 and 02, prediction errors are large, but selecting from 25,000 wells still identifies wells with reserves well above the mean.

**Preliminary choice:** Region 00, because it has the highest potential profit. However, this is a single estimate that does not account for risk, and the selection depends on relatively inaccurate predictions. This choice will be reassessed using bootstrapping to estimate mean profit, the confidence interval, and the probability of a loss.


# Calculating Risk and Profit for Each Region

```python
def bootstrap_profit(target, predictions, n_samples=1000, sample_size=500, state=None):

    '''Estimate the profit distribution using bootstrapping: 1000 samples of 500 wells.'''

    values = []
    for _ in range(n_samples):
        sub_target = target.sample(n=sample_size, replace=True, random_state=state)
        sub_predictions = predictions.loc[sub_target.index]
        values.append(profit(sub_target.reset_index(drop=True),
                            sub_predictions.reset_index(drop=True)))
    return pd.Series(values)

state = np.random.RandomState(12345)
for area, (target_valid, predictions_valid) in regions.items():
    values = bootstrap_profit(target_valid, predictions_valid, state=state)
    print(f'Region {area}: \n',
         f'Mean profit = US$ {values.mean():,.0f}\n',
         f'95% CI = [{values.quantile(0.025):,.0f};{values.quantile(0.975):,.0f}]\n',
         f'Probability of loss = {(values < 0).mean() * 100:.1f}%\n')
    
```

## Evaluating Risk and Profit for the Final Decision

The project requires retaining only regions with a probability of loss below 2.5%. Regions 00 (6.9%) and 02 (6.5%) are well above this threshold, and their confidence intervals include negative values, indicating a risk of losing money. Only **Region 01** meets the criterion, with a **0.7% probability of loss**. It also has **the highest mean profit** (USD 4.61 million) and **an entirely positive confidence interval**, from USD 0.78 million to USD 8.63 million. **I recommend Region 01 for the development of new wells.**

This choice **differs from the choice in step 4.3**, where Region 00 appeared to be the best option, with a potential profit of USD 33.2 million compared with USD 24.2 million for Region 01. In that calculation, the best 200 wells were selected from the 25,000 wells in the validation set, allowing a highly selective choice. In bootstrapping, which represents the exploration of 500 sites, the 200 wells are selected from just 500 sites, or 40% of the sites studied, making the selection much less selective. In this scenario, the less accurate models (R² of 0.28 for Region 00 and 0.21 for Region 02) frequently select low-reserve wells, reducing profit and making it more variable. In contrast, the model for Region 01 is almost exact (RMSE of 0.89), so even a less selective process identifies the right wells, producing stable profits and low risk. Model accuracy therefore has more influence on the decision than the region's maximum reserve volume.

Because the data is synthetic, Region 01's almost perfect performance would probably not be replicated in a real oil field. This limitation should be considered before making an investment decision.
