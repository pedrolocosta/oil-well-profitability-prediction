### Construindo o raciocínio

* esta etapa só foi mantida por entender que pode ser uma parte da requisição das instruções do projeto, normalmente ela seria retirada como foi nas demais.

```python
# dividindo colunas de caracteristicas da coluna objetivo no df alvo
features = df0.drop(['id','product_in_thousands'], axis=1)
target = df0['product_in_thousands']
```

```python
# dividindo os dados em um conjunto de treinamento e um conjunto de validação em uma proporção de 75:25
features_train, features_valid, target_train, target_valid = train_test_split(
    features, target, test_size=0.25, random_state=12345
)
```

```python
# Organizando um df com os dados do conjunto de validação
df0_lr_valid = features_valid.copy()
df0_lr_valid['target'] = target_valid
df0_lr_valid['predictions_valid'] = predictions_valid
```

## Colocando todo o raciocínio em uma função

```python
def train_and_evaluate(df, region, random_state=12345):

    '''Treinando o modelo de regressão linear e avaliando-a com um conjunto de validação'''

    # dividindo colunas de caracteristicas da coluna objetivo no df alvo
    features = df.drop(['id','product_in_thousands'], axis=1)
    target = df['product_in_thousands']

    # dividindo os dados em um conjunto de treinamento e um conjunto de validação em uma proporção de 75:25
    features_train, features_valid, target_train, target_valid = train_test_split(
        features, target, test_size=0.25, random_state=random_state)
    
    # colocando apenas o conjunto de treinamento em escala
    scaler = StandardScaler().fit(features_train)

    # inicializando e treinando o modelo
    model = LinearRegression()
    model.fit(scaler.transform(features_train), target_train)

    # predições para o conj. validação mantendo o índice original (associar as predições com os valores verdadeiros)
    predictions_valid = pd.Series(
        model.predict(scaler.transform(features_valid)),
        index=target_valid.index,
        name='predictions_valid',
    )

    # Organizando um df com os dados do conjunto de validação
    df_lr_valid = features_valid.copy()
    df_lr_valid['target_valid'] = target_valid
    df_lr_valid['predictions_valid'] = predictions_valid
    
    # avaliando o desempenho do modelo
    # calculando a reserva média real na região
    mean_region = df['product_in_thousands'].mean()
    # calculando o volume médio previsto
    predictions_valid_mean = predictions_valid.mean()
    # calculando o REQM do modelo
    rmse = mean_squared_error(target_valid, predictions_valid) ** 0.5
    # calculando o coeficiente de determinação
    r2 = r2_score(target_valid, predictions_valid)
    # calculando os desvios padrão
    target_valid_std = target_valid.std()
    predictions_valid_std = predictions_valid.std()
    
    # imprimindo os resultados da validação
    print(f'Avaliando o desempenho do modelo para a Região {region}:\n')
    print(f'Reserva média real da região {region}: {mean_region:.2f} milhares de barris')
    print(f'Reserva média prevista: {predictions_valid_mean:.2f} milhares de barris')
    print(f'REQM do modelo: {rmse:.2f}')
    print(f'R² do modelo: {r2:.2f}')
    print(
        f'Desvio padrão dos valores reais: {target_valid_std:.2f}', 
        f'| Desvio padrão dos valores preditos {predictions_valid_std:.2f}'
    )
    print('\nDataFrame do conjunto validação:\n', df_lr_valid.head())

    return target_valid, predictions_valid
    
```

## Aplicando a função nas Regiões

### Região 00

```python
# aplicando a função de treinamento e avaliação na região 00
target_valid_0, predictions_valid_0 = train_and_evaluate(df0, '00')
```

### Região 01

```python
# aplicando a função de treinamento e avaliação na região 01
target_valid_1, predictions_valid_1 = train_and_evaluate(df1, '01')
```

### Região 02

```python
# aplicando a função de treinamento e avaliação na região 02
target_valid_2, predictions_valid_2 = train_and_evaluate(df2, '02')
```

# Calculando o lucro de um conjunto do melhores poços de petróleo

## Criando uma função para realizar o cálculo nas regiões

```python
def profit(target, predictions, count=total_well):

    ''' Calcula o lucro dos `count` poços com as maiores predições do modelo.'''

    # índices dos poços com as maiores predições
    best_wells = predictions.sort_values(ascending=False).head(count)

    # encontrando o volume real destes poços (milhares de barris)
    selected = target.loc[best_wells.index]

    # "lucro" = receita dos poços selecionados - investimento
    return selected.sum() * revenue_uni - investiment
```

## Aplicando a função para cada região

```python
profits = {}
top_predictions = {}
for area, (target_valid, predictions_valid) in regions.items():
    # obtendo o lucro para os 200 melhores poço preditos cada região
    profits[area] = profit(target_valid, predictions_valid)
    
    # obtendo as 200 melhores predições de poço por região
    top_predictions[area] = predictions_valid.sort_values(ascending=False).head(total_well)
    
    # obtendo o volume real dos 200 melhores poços por região
    real_volume = target_valid.loc[top_predictions[area].index].sum()
    
    # obtendo a média de unidades por poço em cada região
    mean_well = target_valid.loc[top_predictions[area].index].mean()

    # imprimindo resultado resumido
    print(f'Região {area}:', '\n',
          f'Média de unidades por poço {mean_well:.2f}', '\n',
          f'Volume total de produção = {real_volume:,.0f} unidades de barris', '\n',
          f'Lucro potencial = US$ {profits[area]:,.0f}', '\n')
```

Esta escolha será confrontada com o bootstrapping, que estima o lucro médio, o intervalo de confiança e a probabilidade de prejuízo.

# Calculando os riscos e lucros para cada região

```python
def bootstrap_profit(target, predictions, n_samples=1000, sample_size=500, state=None):

    '''Distribuição de lucros com bootsrapping: 1000 amostras de 500 poços'''

    values = []
    for _ in range(n_samples):
        sub_target = target.sample(n=sample_size, replace=True, random_state=state)
        sub_predictions = predictions.loc[sub_target.index]
        values.append(profit(sub_target.reset_index(drop=True),
                            sub_predictions.reset_index(drop=True)))
    return pd.Series(values)

state = np.random.RandomState(12345)
for area, (target_valid, predictions_valid) in regions.items():
    values = bootstrap_profit(target_valid, predictions_valid, state=state)
    print(f'Região {area}: \n',
         f'Lucro médio = US$ {values.mean():,.0f}\n',
         f'IC 95% = [{values.quantile(0.025):,.0f};{values.quantile(0.975):,.0f}]\n',
         f'Risco de prejuízo = {(values < 0).mean() * 100:.1f}%\n')
    
```

## Avaliando os riscos e lucros para a decisão final

O projeto exige manter apenas regiões com risco de prejuízo inferior a 2,5%. As Regiões 00 (6,9%) e 02 (6,5%) ficam bem acima desse limite, e o intervalo de confiança de ambas inclui valores negativos, ou seja, há chance real de perder dinheiro. Somente a **Região 01** atende ao critério, com `risco de 0,7%`, e ela também tem `o maior lucro médio` (USD 4,61 milhões) e `um intervalo de confiança totalmente positivo` ( de USD 0,78 milhão a USD 8,63 milhões). **Recomendo a Região 01 para o desenvolvimento dos novos poços.**

Essa escolha **não corresponde à da etapa 4.3** em que a Região 00 parecia a melhor (lucros potencial de USD 33,2 milhões contra USD 24,2 milhões da Região 01). Naquele cálculo os 200 melhores poços eram escoolhidos entre os 25 mil do conjunto de validação, uma seleção muito rigorosa. No bootstrapping, que representa o estudo real dos 500 pontos, os 200 poços são escolhidos entre apenas 500, ou seja, 40% dos pontos estudados, e a seleção é muito menos rigorosa. Nesse cenário, modelos pouco precisos (R² de 0,28 na Região 00 e de 0,21 na Região 02) escolhem poços ruins com frequência, o que reduz e torna instável o lucro. Já o modelo da Região 01 é quase exato (REQM de 0,89), então mesmo uma seleção menos rigorosa encontra os poços certos, com lucros estáveis e baixo risco. Em resumo, a precisão do modelo pesa mais na decisão do que o volume máximo de reservas da região.

Como os dados são sintéticos, o desempenho quase perfeito da Região 01 provavelmente não se repetiria em um campo real, o que deve ser considerado antes de uma decisão de investimento.
